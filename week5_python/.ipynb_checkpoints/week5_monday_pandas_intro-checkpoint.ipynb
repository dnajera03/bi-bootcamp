```python
sec_games = df[df['home_conference'] == 'SEC']
print(sec_games.shape)
sec_games.head()
```

```python
sec_recent = df[(df['home_conference'] == 'SEC') & (df['season'] >= 2018)]
print(sec_recent.shape)
sec_recent.head()
```

```python
df[['season', 'home_team', 'away_team', 'attendance']].head()
```

```python
df.sort_values('attendance', ascending=False).head(10)
```

```python
(df[(df['home_conference'] == 'SEC') & (df['season'] >= 2018)] 
 [['season', 'home_team', 'away_team', 'attendance']]
 .sort_values('attendance', ascending=False)
 .head(10)
)
```

```python
df.groupby('home_conference')['attendance'].mean()
```

```python
df.groupby('home_conference').agg(
    avg_attendance=('attendance', 'mean'),
    total_attendance=('attendance', 'sum'),
    game_count=('attendance', 'count'),
    max_attendance=('attendance', 'max')
).sort_values('avg_attendance', ascending=False)
```

```python
df[df['season'] >= 2018].groupby(['home_conference','season'])['attendance'].mean().head(20)
```

```python
conference_info = pd.DataFrame({
    'conference':['SEC', 'Big Ten', 'ACC', 'Big 12', 'Pac-12'],
    'tier':['Power 5', 'Power 5', 'Power 5', 'Power 5', 'Power 5'],
    'founded':[1932, 1896, 1953, 1996, 1959]
})
conference_info
```

```python
df_with_tier = df.merge(conference_info, left_on='home_conference', right_on='conference', how='left')
df_with_tier[['home_team', 'home_conference','tier','founded']].head(10)
```

```python
df_with_tier['tier'].value_counts(dropna=False)
```

```python
pd.pivot_table(
    df[df['season'].between(2018,2024)],
    values='attendance',
    index='season',
    columns='home_conference',
    aggfunc='mean'
).round(0)
```
